```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler, PolynomialFeatures
from sklearn.model_selection import GridSearchCV
from sklearn.linear_model import Lasso
import plotly.express as px
```

Load Dataset

```python
df = pd.read_csv("CleanPreprocessedExoplanet2.csv")
df = df.drop("Unnamed: 0", axis=1)
df.head()
```

Split into features and targets

```python
X = df.drop("log radius", axis=1)
y = df["log radius"]
```

Create polynomial features

```python
degree = 3
features = PolynomialFeatures(degree, include_bias=False).set_output(transform="pandas")
Xe = features.fit_transform(X)
Xe.shape
```

Standardize features

```python
stnd = StandardScaler().set_output(transform="pandas")
Xe = stnd.fit_transform(Xe)
```

Apply lasso grid search

```python
grid = {"alpha": 10**np.linspace(-6,6)}
lasso = Lasso()
lassoCV  = GridSearchCV(lasso, grid)
```

```python
lassoCV.fit(Xe, y)
```

Save features where the coeficients are above $10^{-8}$ to a csv file

```python
Xe[Xe.columns[np.abs(lassoCV.coefs_) > 1e-8]].to_csv("LassoFeatures2.csv")
```
